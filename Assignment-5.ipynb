## 1. Statistical Derivation: Bias in the MLE Covariance Matrix

The Maximum Likelihood Estimator (MLE) for the population covariance matrix $\boldsymbol{\Sigma}$ is defined as:

$$\widehat{\boldsymbol{\Sigma}}_{\text{MLE}} = \frac{1}{n} \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T$$

where $\widehat{\boldsymbol{\mu}}_n = \frac{1}{n}\sum_{i=1}^n \mathbf{X}_i$ denotes the empirical sample mean vector.

To analyze the estimator's bias, we decompose the deviation of each observation from the sample mean by introducing the true population mean $\boldsymbol{\mu}$:

$$\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n = (\mathbf{X}_i - \boldsymbol{\mu}) - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})$$

Substituting this identity back into the outer product sum yields:

$$\sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T = \sum_{i=1}^n \left[ (\mathbf{X}_i - \boldsymbol{\mu}) - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \right] \left[ (\mathbf{X}_i - \boldsymbol{\mu}) - (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \right]^T$$

Expanding this product term-by-term distributes the summation across four distinct components:

$$= \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T - \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T - \sum_{i=1}^n (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T + \sum_{i=1}^n (\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T$$

We can simplify the cross-product terms by noting that $\sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu}) = n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})$. Substituting this into the middle terms gives:

$$= \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T - n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T - n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T + n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T$$

Consolidating the scalar multipliers reduces the expression to:

$$\sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T = \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T - n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T$$

Applying the expectation operator $\mathbb{E}[\cdot]$ to the scaled MLE equation, we get:

$$\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] = \frac{1}{n} \sum_{i=1}^n \mathbb{E}[(\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T] - \mathbb{E}[(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T]$$

Using the fundamental definitions of population covariance and the variance of the sample mean:
* $\mathbb{E}[(\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T] = \boldsymbol{\Sigma}$
* $\mathbb{E}[(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T] = \text{Var}(\widehat{\boldsymbol{\mu}}_n) = \frac{1}{n}\boldsymbol{\Sigma}$

Substituting these moments back into our expectation formula yields:

$$\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] = \frac{1}{n} (n\boldsymbol{\Sigma}) - \frac{1}{n}\boldsymbol{\Sigma} = \boldsymbol{\Sigma} - \frac{1}{n}\boldsymbol{\Sigma} = \left(\frac{n-1}{n}\right)\boldsymbol{\Sigma}$$

> **Conclusion on Bias:** Because $\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] \neq \boldsymbol{\Sigma}$, the MLE systematically underestimates the true population covariance. This downward bias arises because the sample points are naturally closer to their own calculated mean $\widehat{\boldsymbol{\mu}}_n$ than to the true population mean $\boldsymbol{\mu}$, artificially restricting the observed geometric spread.

---

### Correcting Bias via Bessel's Correction

To construct an unbiased estimator ($\mathbf{S}$), we scale the MLE matrix by the reciprocal of the bias factor, $\frac{n}{n-1}$:

$$\mathbf{S} = \frac{n}{n-1} \widehat{\boldsymbol{\Sigma}}_{\text{MLE}} = \frac{n}{n-1} \left( \frac{1}{n} \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T \right) = \frac{1}{n-1} \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T$$

Verifying the expectation confirms that the bias has been successfully eliminated:

$$\mathbb{E}[\mathbf{S}] = \frac{n}{n-1} \mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] = \frac{n}{n-1} \left(\frac{n-1}{n}\right)\boldsymbol{\Sigma} = \boldsymbol{\Sigma}$$

## 2. Operational Error Diagnostics & Threshold Adjustments

In structural health monitoring and anomaly detection, decision errors are classified under two main frameworks:

* **Type I Error ($\alpha$ - False Alarm Rate):** Occurs when the diagnostic model erroneously rejects the null hypothesis ($H_0$). The system flags a completely healthy structural asset as damaged or unstable, resulting in unnecessary downtime and maintenance costs.
* **Type II Error ($\beta$ - Missed Detection Rate):** Occurs when the diagnostic model fails to reject $H_0$ despite a true underlying change. A structural defect or fracture exists ($H_1$ is true), but the system categorizes the state as nominal, risking catastrophic asset failure.

### Consequences of a Strict Significance Threshold ($\alpha = 0.0001$)

Setting an ultra-conservative significance level alters the sensitivity of the entire diagnostic framework in two major ways:

1. **Collapse of Statistical Power ($1-\beta$):** By minimizing $\alpha$ to near zero, the critical decision boundary is pushed deep into the extreme tails of the baseline distribution. This drastically reduces the probability of a false alarm, but makes it highly difficult for early-stage or minor structural faults to cross the decision threshold. Consequently, the missed detection rate ($\beta$) spikes, causing the **statistical power ($1-\beta$) to drop sharply**.

2. **Expansion of Confidence Ellipsoids:** The confidence level governing the multi-dimensional boundary scales to $1-\alpha = 0.9999$. Geometrically, this requires the **volume of the healthy operational ellipsoid to expand significantly** to capture nearly all potential baseline variations. Because this boundary swells outwards, the model becomes insensitive to localized structural degradations, treating anomalous data shifts as normal variations.

## 1. Slutsky's Theorem: Core Formulation

Let $\{X_n\}$ and $\{Y_n\}$ be sequences of random variables (or vectors). If $X_n$ converges in distribution to a random variable $X$ ($X_n \xrightarrow{d} X$), and $Y_n$ converges in probability to a constant scalar, vector, or matrix $c$ ($Y_n \xrightarrow{p} c$), then the following asymptotic relationships hold:

* **Additive Stability:** $$X_n + Y_n \xrightarrow{d} X + c$$
* **Multiplicative Scaling:** $$X_n Y_n \xrightarrow{d} cX$$
* **Divisive Scaling:** $$X_n / Y_n \xrightarrow{d} X/c \quad (\text{provided } c \neq 0)$$

## 2. Asymptotic Justification for Covariance Substitution

In multivariate statistical diagnostics, we leverage Slutsky's Theorem to substitute an unknown population covariance matrix $\boldsymbol{\Sigma}$ with its empirical sample counterpart $\mathbf{S}$. The formal mathematical justification proceeds as follows:

### Step 1: Baseline Convergence via the CLT
By the Lindeberg-Lévy Multivariate Central Limit Theorem (CLT), the standardized sample mean vector asymptotically approaches a multivariate normal distribution:

$$\sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \xrightarrow{d} \mathcal{N}(\mathbf{0}, \boldsymbol{\Sigma})$$

Multiplying by the inverse matrix square root $\boldsymbol{\Sigma}^{-1/2}$ normalizes the space to a standard Gaussian profile:

$$\boldsymbol{\Sigma}^{-1/2} \left[ \sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \right] \xrightarrow{d} \mathcal{N}(\mathbf{0}, \mathbf{I})$$

### Step 2: Consistency of the Sample Estimator
By the Weak Law of Large Numbers (WLLN), the unbiased sample covariance matrix $\mathbf{S}$ is a consistent estimator of the population covariance matrix $\boldsymbol{\Sigma}$. By applying the continuous mapping theorem, its inverse square root converges in probability to a fixed constant matrix:

$$\mathbf{S} \xrightarrow{p} \boldsymbol{\Sigma} \implies \mathbf{S}^{-1/2} \xrightarrow{p} \boldsymbol{\Sigma}^{-1/2}$$

### Step 3: Application of Slutsky's Theorem
We now multiply the sequence converging in distribution (from Step 1) by the sequence converging in probability to a constant matrix (from Step 2). According to Slutsky's Theorem, this combination preserves the target distribution:

$$\mathbf{S}^{-1/2} \left[ \sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \right] \xrightarrow{d} \boldsymbol{\Sigma}^{-1/2} \cdot \mathcal{N}(\mathbf{0}, \boldsymbol{\Sigma}) = \mathcal{N}(\mathbf{0}, \mathbf{I})$$

## 3. Engineering & Operational Implications

Because the statistic converges asymptotically to a standard normal distribution under the sample metric, we can map the localized coordinate parameters back to the empirical domain for large sample sizes ($n \to \infty$):

$$\sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \sim \mathcal{N}(\mathbf{0}, \mathbf{S}) \implies \widehat{\boldsymbol{\mu}}_n \sim \mathcal{N}\left(\boldsymbol{\mu}, \frac{1}{n}\mathbf{S}\right)$$

> **Operational Significance:** This mathematical framework validates substituting the sample covariance matrix $\mathbf{S}$ for the unknown population matrix $\boldsymbol{\Sigma}$. It guarantees that the geometry of the asymptotic diagnostic envelope (such as a Mahalanobis distance threshold used for anomaly detection) remains stable, theoretically sound, and free from structural distortion as the sample size grows.

## 1. Mathematical Framework for First Moment Homogeneity

To verify the weak i.i.d. assumption, we test the null hypothesis that the mean vector remains constant across all $g = 5$ non-overlapping temporal blocks:

$$H_0: \boldsymbol{\mu}_1 = \boldsymbol{\mu}_2 = \dots = \boldsymbol{\mu}_g$$
$$H_1: \boldsymbol{\mu}_j \neq \boldsymbol{\mu}_k \quad \text{for at least one pair } (j, k)$$

The structural deviations are partitioned into the **Within-Group Sum of Squares and Cross-Products (SSCP) matrix ($\mathbf{W}$)** and the **Between-Group SSCP matrix ($\mathbf{B}$)**:

$$\mathbf{W} = \sum_{j=1}^g \sum_{i=1}^{n_j} (\mathbf{X}_{ij} - \widehat{\boldsymbol{\mu}}_j)(\mathbf{X}_{ij} - \widehat{\boldsymbol{\mu}}_j)^T$$

$$\mathbf{B} = \sum_{j=1}^g n_j (\widehat{\boldsymbol{\mu}}_j - \widehat{\boldsymbol{\mu}})(\widehat{\boldsymbol{\mu}}_j - \widehat{\boldsymbol{\mu}})^T$$

where $\widehat{\boldsymbol{\mu}}_j$ is the local sample mean vector of block $j$, and $\widehat{\boldsymbol{\mu}}$ is the grand mean vector of the entire dataset.

def run_moment_homogeneity_test(dataframe, channels, group_col):
    """
    Computes Wilks' Lambda and applies Bartlett's Chi-Square approximation
    to check for mean vector drift across temporal groups.
    """
    n = len(dataframe)
    m = len(channels)
    g = dataframe[group_col].nunique()

$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbb{E}\left[ \mathbf{P}^T \widetilde{\mathbf{X}}_i \widetilde{\mathbf{X}}_i^T \mathbf{P} \right]$$

$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{P}^T \mathbb{E}[\widetilde{\mathbf{X}}_i \widetilde{\mathbf{X}}_i^T] \mathbf{P} = \mathbf{P}^T \boldsymbol{\Sigma} \mathbf{P}$$

* **Statistical:** Because $\mathbf{\Lambda}$ is strictly diagonal, $\text{Cov}(Z_{i,j}, Z_{i,k}) = 0$ for $j \neq k$. The transformation completely **de-correlates** the features, eliminating linear redundancy.

Evaluating the squared Euclidean norm via vector expansion:

$$\|\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n\|^2 = \left( \widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k} \right)^T \left( \widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k} \right)$$

$$= \mathbf{z}_{i,k}^T \widehat{\mathbf{P}}_k^T \widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \mathbf{z}_{i,k}^T \widehat{\mathbf{P}}_k^T \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k} + \mathbf{z}_{i,m-k}^T \widehat{\mathbf{P}}_{m-k}^T \widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \mathbf{z}_{i,m-k}^T \widehat{\mathbf{P}}_{m-k}^T \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k}$$

Applying intra-subspace orthonormality ($\widehat{\mathbf{P}}_k^T \widehat{\mathbf{P}}_k = \mathbf{I}$, $\widehat{\mathbf{P}}_{m-k}^T \widehat{\mathbf{P}}_{m-k} = \mathbf{I}$) and inter-subspace orthogonality ($\widehat{\mathbf{P}}_k^T \widehat{\mathbf{P}}_{m-k} = \mathbf{0}$), the cross-product terms vanish:

$$\|\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n\|^2 = \mathbf{z}_{i,k}^T \mathbf{I} \mathbf{z}_{i,k} + \mathbf{0} + \mathbf{0} + \mathbf{z}_{i,m-k}^T \mathbf{I} \mathbf{z}_{i,m-k} = \|\mathbf{z}_{i,k}\|^2 + \|\mathbf{z}_{i,m-k}\|^2$$

### SHM Diagnostic Strategy

* **Hotelling's $T^2 = \mathbf{z}_{i,k}^T \mathbf{\Lambda}_k^{-1} \mathbf{z}_{i,k}$** monitors variations **within** the principal subspace ($\widehat{\mathbf{P}}_k$).
* **Residual $Q = \mathbf{e}_i^T \mathbf{e}_i$** monitors variations **orthogonal** to the principal subspace ($\widehat{\mathbf{P}}_{m-k}$).

| Event Type | Most Sensitive Statistic | Engineering Justification |
| :--- | :--- | :--- |
| **Environmental Load Anomaly** (e.g., Extreme Wind) | **Hotelling's $T^2$** | The underlying structure is undamaged, keeping sensor cross-correlations intact. The system simply experiences large-amplitude excursions inside the established nominal operational subspace. |
| **Internal Structural Fracture** (e.g., Fatigue Crack) | **Residual $Q$ (SPE)** | The physical crack breaks historical sensor correlation patterns and alters load paths. This new geometric behavior cannot be captured by $\widehat{\mathbf{P}}_k$, causing data to project directly into the residual space. |

### Proof of the Fundamental Equation of Factor Analysis

The model posits the linear generative structure for the normalized, zero-mean baseline feature vector $Z_i \in \mathbb{R}^m$:

$$Z_i = \Lambda F_i + \epsilon_i$$

To find the population covariance matrix $R$, we evaluate the expected value of the outer product $E[Z_i Z_i^T]$:

$$R = E[Z_i Z_i^T] = E[(\Lambda F_i + \epsilon_i)(\Lambda F_i + \epsilon_i)^T]$$

Expanding the terms inside the expectation operator yields:

$$R = E[\Lambda F_i F_i^T \Lambda^T + \Lambda F_i \epsilon_i^T + \epsilon_i F_i^T \Lambda^T + \epsilon_i \epsilon_i^T]$$

By the linearity of the expectation operator, this splits into individual terms:

$$R = \Lambda E[F_i F_i^T] \Lambda^T + \Lambda E[F_i \epsilon_i^T] + E[\epsilon_i F_i^T] \Lambda^T + E[\epsilon_i \epsilon_i^T]$$

Substituting the structural model constraints:
1. Orthogonal common factors: $E[F_i F_i^T] = I_{k \times k}$
2. Independence of latent factors and errors: $E[\epsilon_i F_i^T] = 0 \implies E[F_i \epsilon_i^T] = 0$
3. Diagonalized measurement noise covariance: $E[\epsilon_i \epsilon_i^T] = \Psi$

Substituting these identities back into the expanded equation:

$$R = \Lambda (I_{k \times k}) \Lambda^T + \Lambda (0) + (0) \Lambda^T + \Psi$$

$$R = \Lambda \Lambda^T + \Psi$$

def varimax_rotation(Lambda, max_iter=500, tol=1e-6):
    """
    Performs an orthogonal Varimax rotation on the factor loading matrix Lambda.
    """
    p, k = Lambda.shape
    if k < 2:
        return Lambda, np.eye(k)
    
    T = np.eye(k)
    X = Lambda.copy()
    
    for i in range(max_iter):
        X_old = X.copy()
        d = np.diag(np.sum(X**2, axis=0))
        u = X**3 - (1.0 / p) * X @ d
        
        U, S, Vt = np.linalg.svd(X.T @ u)
        T_next = U @ Vt
        
        X = Lambda @ T_next
        if np.allclose(X, X_old, atol=tol):
            T = T_next
            break
        T = T_next
        
    return X, T

````python
import numpy as np
import pandas as pd
from typing import Optional, Sequence, Dict, Any
from sklearn.decomposition import FactorAnalysis
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# =====================================================================
# COMBINED SUBSPACE ENGINES (PCA & FA PIPELINES)
# =====================================================================

class DualSubspaceDiagnosticsEngine:
    def __init__(self, df: pd.DataFrame):
        self.df = df

    def compute_empirical_pca(self, columns: Optional[Sequence[str]] = None, show_plot: bool = True) -> Dict[str, Any]:
        """
        Operationalizes the geometric de-correlation framework via PCA.
        Decomposes the unbiased empirical covariance matrix S into its orthogonal
        basis P_hat and renders a clean 2x3 optimization and feature contribution dashboard.
        """
        if self.df is None:
            raise ValueError("Error: No data loaded.")

        if columns is None:
            target_cols = self.df.select_dtypes(include=[np.number]).columns.tolist()
            if 'count' in target_cols: target_cols.remove('count')
            target_cols = list(target_cols)
        else:
            target_cols = list(columns)

        X = self.df[target_cols].copy().dropna().values
        n, m = X.shape

        if n <= m:
            raise ValueError("Sample size n must be larger than feature dimensions m to isolate principal components.")

        # 1. Centering
        mu_hat = np.mean(X, axis=0)
        X_centered = X - mu_hat

        # 2. Compute Unbiased Sample Covariance Matrix S
        S_matrix = np.cov(X, rowvar=False, ddof=1)

        # 3. Spectral Decomposition
        eigenvalues, eigenvectors = np.linalg.eigh(S_matrix)
        idx = np.argsort(eigenvalues)[::-1]
        lambda_hat = eigenvalues[idx]
        P_hat = eigenvectors[:, idx]
        lambda_hat = np.clip(lambda_hat, a_min=1e-15, a_max=None)

        # 4. Variance Information Calculations
        total_variance = np.sum(lambda_hat)
        if total_variance == 0:
            raise ValueError("Total variance is zero.")

        explained_variance_ratio = lambda_hat / total_variance
        cumulative_variance_ratio = np.cumsum(explained_variance_ratio)
        unexplained_variance_ratio = 1.0 - cumulative_variance_ratio

        # 5. Map Realizations to Principal Component Scores
        Z_scores = np.dot(X_centered, P_hat)
        S_Z = np.cov(Z_scores, rowvar=False, ddof=1)

        # 6. Evaluate Statistical Distance Metrics (T^2 and Q)
        k_range = np.arange(1, m)
        mean_T2_vs_k = []
        mean_Q_vs_k = []
        T2_matrix = np.zeros((n, len(k_range)))
        Q_matrix = np.zeros((n, len(k_range)))

        for idx_k, k in enumerate(k_range):
            Z_k = Z_scores[:, :k]
            lambda_k = lambda_hat[:k]
            T2_samples = np.sum((Z_k ** 2) / lambda_k, axis=1)
            T2_matrix[:, idx_k] = T2_samples
            mean_T2_vs_k.append(np.mean(T2_samples))

            Z_residual = Z_scores[:, k:]
            Q_samples = np.sum(Z_residual ** 2, axis=1)
            Q_matrix[:, idx_k] = Q_samples
            mean_Q_vs_k.append(np.mean(Q_samples))

        print(f"\n--- Operationalizing Geometric De-correlation Layer (PCA) ---")
        print(f"Decomposing structural space of {m} features using {n} samples.")
        print(f"Total System Variance (Trace[S]): {total_variance:.4f}")

        # 7. Enhanced 2x3 Diagnostic Dashboard
        if show_plot:
            pc_labels = [f"PC {i+1}" for i in range(m)]
            k_labels = [f"k={k}" for k in k_range]
            sensor_labels = target_cols

            fig = make_subplots(
                rows=2, cols=3,
                horizontal_spacing=0.18,
                vertical_spacing=0.28,
                subplot_titles=(
                    "Feature Loading Matrix |P_hat|",
                    "Component Values (Eigenvalues λ)",
                    "Information Profile (Explained Var.)",
                    "Residual Space (Unexplained Var.)",
                    "Mean Hotelling's T² vs Subspace Size k",
                    "Mean Q Statistic (SPE) vs Subspace Size k"
                )
            )

            # --- ROW 1, COL 1: PCA Loading Heatmap (Left-anchored Colorbar) ---
            fig.add_trace(
                go.Heatmap(
                    z=np.abs(P_hat),
                    x=pc_labels,
                    y=sensor_labels,
                    colorscale='YlOrRd',
                    colorbar=dict(
                        title="Loading Weight",
                        x=-0.12,          # Perfectly isolates colorbar to the left side
                        len=0.38,         # Fits height bounds of row 1
                        y=0.78,           # Centers vertically inside row 1
                        yanchor="middle",
                        xanchor="right",
                        titleside="top"
                    ),
                    showscale=True,
                    showlegend=False
                ),
                row=1, col=1
            )
            fig.update_xaxes(title_text="Principal Axes", row=1, col=1)

            # --- ROW 1, COL 2: Absolute Eigenvalues ---
            fig.add_trace(go.Bar(x=pc_labels, y=lambda_hat, name="Eigenvalue (λ_j)", marker=dict(color='#1f77b4', line=dict(color='black', width=0.5)), legendgroup="eigen"), row=1, col=2)
            fig.update_yaxes(title_text="Variance Magnitude", row=1, col=2)
            fig.update_xaxes(title_text="Principal Axes", row=1, col=2)

            # --- ROW 1, COL 3: Variance Profiles ---
            fig.add_trace(go.Bar(x=pc_labels, y=explained_variance_ratio * 100, name="Marginal Explained", marker=dict(color='#ff7f0e', opacity=0.75), legendgroup="expl"), row=1, col=3)
            fig.add_trace(go.Scatter(x=pc_labels, y=cumulative_variance_ratio * 100, mode='lines+markers', name='Cumulative Captured', line=dict(color='#d62728', width=2.5, dash='dash'), legendgroup="expl"), row=1, col=3)
            fig.update_yaxes(title_text="Captured Structure (%)", range=[-2, 105], row=1, col=3)
            fig.update_xaxes(title_text="Principal Axes", row=1, col=3)

            # --- ROW 2, COL 1: Remaining Noise ---
            fig.add_trace(go.Bar(x=pc_labels, y=unexplained_variance_ratio * 100, name="Remaining Noise", marker=dict(color='#2ca02c', line=dict(color='black', width=0.5)), legendgroup="noise"), row=2, col=1)
            fig.update_yaxes(title_text="Excluded Info (%)", range=[-2, 105], row=2, col=1)
            fig.update_xaxes(title_text="Principal Axes", row=2, col=1)

            # --- ROW 2, COL 2: Hotelling's T^2 ---
            fig.add_trace(go.Scatter(x=k_labels, y=mean_T2_vs_k, mode='lines+markers', name='Mean T²', line=dict(color='#9467bd', width=2.5), marker=dict(size=6, symbol='diamond'), legendgroup="t2"), row=2, col=2)
            fig.update_yaxes(title_text="Average T² Metric", row=2, col=2)
            fig.update_xaxes(title_text="Truncation Cutoff (k)", row=2, col=2)

            # --- ROW 2, COL 3: SPE / Q Statistic ---
            fig.add_trace(go.Scatter(x=k_labels, y=mean_Q_vs_k, mode='lines+markers', name='Mean Q (SPE)', line=dict(color='#e377c2', width=2.5), marker=dict(size=6, symbol='square'), legendgroup="q_stat"), row=2, col=3)
            fig.update_yaxes(title_text="Average Residual Energy", row=2, col=3)
            fig.update_xaxes(title_text="Truncation Cutoff (k)", row=2, col=3)

            # Global Unified Layout Configurations
            fig.update_layout(
                title=dict(text="Principal Component Analysis (PCA) Optimization & Feature Loading Dashboard", x=0.5, y=0.97, xanchor="center", yanchor="top"),
                template="plotly_white",
                showlegend=True,
                legend=dict(
                    orientation="h",
                    yanchor="bottom",
                    y=1.02,
                    xanchor="center",
                    x=0.5
                ),
                margin=dict(t=150, b=60, l=140, r=80),
                height=750,
                width=1250
            )
            fig.show()

        return {
            "mean_vector": mu_hat, "covariance_matrix_S": S_matrix, "eigenvalues_lambda": lambda_hat,
            "eigenvectors_P": P_hat, "explained_variance_ratio": explained_variance_ratio,
            "cumulative_variance_ratio": cumulative_variance_ratio, "unexplained_variance_ratio": unexplained_variance_ratio,
            "transformed_scores_Z": Z_scores, "score_covariance_diagonal": np.diag(S_Z),
            "features": target_cols, "k_values": k_range, "T2_matrix_vs_k": T2_matrix, "Q_matrix_vs_k": Q_matrix,
            "mean_T2_profile": np.array(mean_T2_vs_k), "mean_Q_profile": np.array(mean_Q_vs_k)
        }

    def compute_empirical_fa(self, k: int, columns: Optional[Sequence[str]] = None, show_plot: bool = True) -> Dict[str, Any]:
        """
        Operationalizes the Factor Analysis latent subspace framework with a clean 2x2 grid.
        Anchors the heatmap colorbar perfectly to the right edge of the first subplot.
        """
        if self.df is None:
            raise ValueError("Error: No data loaded.")

        if columns is None:
            target_cols = self.df.select_dtypes(include=[np.number]).columns.tolist()
            if 'count' in target_cols: target_cols.remove('count')
            target_cols = list(target_cols)
        else:
            target_cols = list(columns)

        X = self.df[target_cols].copy().dropna().values
        n, m = X.shape

        # 1. Standardize realizations to construct Z space
        mu_hat = np.mean(X, axis=0)
        std_hat = np.std(X, axis=0, ddof=1)
        std_hat[std_hat == 0] = 1e-15
        Z = (X - mu_hat) / std_hat
        R_matrix = np.corrcoef(X, rowvar=False)

        # 2. Fit Factor Analysis Model (Varimax Rotation)
        fa = FactorAnalysis(n_components=k, rotation='varimax', random_state=42)
        fa.fit(Z)
        lambda_matrix = fa.components_.T
        psi_diagonal = fa.noise_variance_

        # 3. Calculate Communality (h²) and Uniqueness (φ²) vectors
        communality = np.sum(lambda_matrix**2, axis=1)
        uniqueness = psi_diagonal

        # 4. Map realizations using Thomson's Regression Method
        F_scores = fa.transform(Z)

        print(f"\n--- Operationalizing Latent Subspace Layer (Factor Analysis) ---")
        print(f"Decomposing structural space of {m} sensors into {k} hidden physical factors.")
        print(f"Average System Communality (Shared Subspace Energy): {np.mean(communality)*100:.2f}%")
        print(f"Average System Uniqueness (Localized Noise Floor): {np.mean(uniqueness)*100:.2f}%")

        # 5. Clean 2x2 Dashboard
        if show_plot:
            sensor_labels = target_cols
            factor_labels = [f"Factor {j+1}" for j in range(k)]

            fig = make_subplots(
                rows=2, cols=2, horizontal_spacing=0.24, vertical_spacing=0.28,
                subplot_titles=(
                    "Structural Loadings Matrix |λ_(j,r)|",
                    "Variance Partitioning (Communality vs Uniqueness)",
                    "Sensor Uniqueness Noise Floor (φ²)",
                    "Latent Factor Scores Empirical Variance"
                )
            )

            # --- Subplot 1: Factor Loadings Heatmap (Colorbar repositioned to x=0.42) ---
            fig.add_trace(
                go.Heatmap(
                    z=np.abs(lambda_matrix),
                    x=factor_labels,
                    y=sensor_labels,
                    colorscale='YlOrRd',
                    colorbar=dict(
                        title="Sensitivity",
                        x=-0.1,          # Shifts colorbar out of the far right to center-left
                        len=0.38,         # Scales down length to match row 1 height bounds
                        y=0.78,          # Centers vertically inside row 1
                        yanchor="middle",
                        xanchor="right"
                    ),
                    showscale=True,
                    showlegend=False
                ),
                row=1, col=1
            )
            fig.update_xaxes(title_text="Latent Structures", row=1, col=1)

            # --- Subplot 2: Variance Partitioning ---
            fig.add_trace(go.Bar(y=sensor_labels, x=communality * 100, name="Communality (h² - Shared Structure)", orientation='h', marker=dict(color='#1f77b4'), legendgroup="communality"), row=1, col=2)
            fig.add_trace(go.Bar(y=sensor_labels, x=uniqueness * 100, name="Uniqueness (φ² - Channel Noise)", orientation='h', marker=dict(color='#ff7f0e'), legendgroup="uniqueness_bar"), row=1, col=2)
            fig.update_layout(barmode='stack')
            fig.update_xaxes(title_text="Variance Allocation (%)", range=[0, 100], row=1, col=2)

            # --- Subplot 3: Uniqueness Line Profile ---
            fig.add_trace(go.Scatter(x=sensor_labels, y=uniqueness, mode='lines+markers', name='Uniqueness Profile (φ²)', line=dict(color='#d62728', width=2, dash='dot'), marker=dict(size=8, symbol='x'), legendgroup="uniqueness_line"), row=2, col=1)
            fig.update_yaxes(range=[-0.05, 1.05], row=2, col=1)
            fig.update_xaxes(title_text="Monitored Channels", tickangle=25, row=2, col=1)

            # --- Subplot 4: Latent Factor Scores Empirical Variance ---
            factor_variances = np.var(F_scores, axis=0, ddof=1)
            fig.add_trace(go.Bar(x=factor_labels, y=factor_variances, name="Factor Empirical Variance", marker=dict(color='#2ca02c', line=dict(color='black', width=0.5)), legendgroup="variance"), row=2, col=2)
            fig.update_yaxes(title_text="Variance Level", row=2, col=2)
            fig.update_xaxes(title_text="Latent Vectors", row=2, col=2)

            # Global Layout Updates
            fig.update_layout(
                title=dict(text="Factor Analysis (FA) Latent Subspace Diagnostics Dashboard", x=0.5, y=0.97, xanchor="center", yanchor="top"),
                template="plotly_white",
                showlegend=True,
                legend=dict(
                    orientation="h",
                    yanchor="bottom",
                    y=1.02,
                    xanchor="center",
                    x=0.5,
                    traceorder="normal",
                    itemwidth=40,
                    itemsizing="constant"
                ),
                margin=dict(t=100, b=60, l=80, r=80),
                height=750,
                width=1100
            )
            fig.show()

        return {
            "mean_vector_mu": mu_hat, "std_vector_D": std_hat, "correlation_matrix_R": R_matrix,
            "factor_loadings_lambda": lambda_matrix, "uniqueness_psi": uniqueness,
            "communality_h2": communality, "latent_factor_scores_F": F_scores, "sensors": target_cols
        }

# =====================================================================
# DATA GENERATOR & PIPELINE EXECUTION
# =====================================================================
if __name__ == "__main__":
    # Set execution seed for perfect reproducible properties
    np.random.seed(42)
    n_samples = 2500

    # 1. Generate underlying, independent true physical driving forces (Latent Factors)
    f1 = np.random.normal(0, 1, n_samples)  # Primary structural mode/load
    f2 = np.random.normal(0, 1, n_samples)  # Secondary operational mode

    # 2. Construct the Observable Multi-Sensor Array with targeted noise profiles
    # Sensors 1 and 2 couple strongly to the primary process (f1)
    s1 = 0.85 * f1 + 0.10 * f2 + np.random.normal(0, 0.3, n_samples)
    s2 = 0.80 * f1 + 0.15 * f2 + np.random.normal(0, 0.35, n_samples)

    # Sensor 3 couples strongly to the secondary process (f2)
    s3 = 0.12 * f1 + 0.90 * f2 + np.random.normal(0, 0.25, n_samples)

    # Sensor 4 suffers an extreme electrical/instrumentation fault (massive localized white noise)
    s4 = 0.02 * f1 + 0.05 * f2 + np.random.normal(0, 1.40, n_samples)

    # 3. Compile into the final evaluation DataFrame
    df_asset = pd.DataFrame(
        data=np.vstack([s1, s2, s3, s4]).T,
        columns=['Sensor_1', 'Sensor_2', 'Sensor_3', 'Sensor_4']
    )
    # Instantiate Engine
    engine = DualSubspaceDiagnosticsEngine(df_asset)

    # Run PCA Engine (1x5 Dashboard)
    pca_metrics = engine.compute_empirical_pca(show_plot=True)

    # Run FA Engine (2x2 Dashboard, Retaining k=2 Common Physical Factors)
    fa_metrics = engine.compute_empirical_fa(k=2, show_plot=True)
````

### Solution for Question 1: The Total Variance Illusion

This is the classic trap of unsupervised dimensionality reduction. PCA and Factor Analysis (FA) look at the world through two entirely different mathematical lenses, and Sensor_4 is the perfect example of how PCA can be "fooled" by a loud, isolated signal.

Here is the breakdown of what is actually happening under the hood of your dashboard.

---

## 1. The Physical Nature of Sensor_4 (Uniqueness $\approx$ 98%)

In Factor Analysis, a variable's total variance is split into **communality** (variance shared with other variables) and **uniqueness** (variance specific to that variable plus measurement noise).

$$Total\ Variance = Communality\ (\theta^2) + Uniqueness\ (\psi^2)$$

Because Sensor_4 has a uniqueness value ($\psi^2$) exceeding 98%, its communality is less than 2%. Physically, this tells us two things:

* **Complete Isolation:** Sensor_4 shares virtually zero underlying physical dynamics with the rest of the plant. Whatever it is measuring (e.g., localized ambient temperature, an isolated tank level, or pure sensor degradation), it does not covary with the core process drivers that the other sensors are tracking.
* **Dominant Local Noise / Idiosyncrasy:** The data from Sensor_4 is almost entirely composed of unique variations or independent high-amplitude noise. It is a solo act, completely detached from the "symphony" of the rest of the plant.

---

## 2. How the Noise "Tricks" the PCA Engine

PCA is fundamentally blind to the *source* of variance; it only cares about the *magnitude* of variance.

### The PCA Objective
PCA operates on the total covariance matrix (or correlation matrix) and seeks orthogonal axes that maximize the total global variance. It solves the eigenvalue problem:

$$\mathbf{\Sigma} \mathbf{v} = \lambda \mathbf{v}$$

Where $\mathbf{\Sigma}$ is the total covariance matrix and $\lambda$ represents the eigenvalues (the variance captured by each principal component).

### The Illusion
If Sensor_4 has massive localized volatility (noise), its diagonal entry in the covariance matrix ($\sigma_4^2$) will be exceptionally large.

Because PCA allocates its first few principal components (PCs) to capture the largest chunks of total variance, **it will violently steer a principal component directly toward Sensor_4's axis** just to capture that massive localized energy.

* **Inflation:** The eigenvalue ($\lambda$) for PC 1 or PC 2 gets artificially inflated because it is absorbing the raw, unshared noise of Sensor_4.
* **The Deception:** Your dashboard shows PC 1 and PC 2 capturing 45% and 30% of the variance, making it look like a highly structured, multi-component system. In reality, a massive portion of that "structure" is just PCA faithfully tracking the meaningless, high-amplitude erraticism of a single isolated sensor.

---

## 3. The Major Risk in Plant Anomaly Detection

Relying solely on this PCA configuration to build an automated plant anomaly detection framework introduces severe systemic risks:

### Blindness to True Systemic Anomalies
Because PC 1 and PC 2 have been hijacked by the noise of Sensor_4, the true underlying process dynamics (the shared variance among the other sensors) are squeezed into the lower, neglected components (PC 3, PC 4, etc.). If a genuine, dangerous systemic anomaly begins to develop across the plant, the top PCs may completely miss it because they are too busy tracking Sensor_4.

### A Flood of False Alarms
Since the top PCs are heavily weighted on Sensor_4, any benign local spike, electrical glitch, or routine maintenance noise on Sensor_4 will trigger a massive shift in your PCA space. This will blow past your Hotelling’s $T^2$ or SPE (Squared Prediction Error) control limits, triggering false emergency shutdowns or alarm fatigue for the operators.

### The Fix
To prevent this "Total Variance Illusion," an engineer should either:
1. **Use Factor Analysis** for the anomaly framework, as it models the shared covariance ($\mathbf{\Sigma} - \mathbf{\Psi}$) and ignores the uniqueness diagonal.
2. **Pre-filter the data**, stripping out or heavily penalizing variables like Sensor_4 that exhibit high uniqueness before passing the matrix to a PCA engine.

### Solution for Question 2: Decoupling Structural Loading via Rotation

* **Part 1 (Mathematical Jumble vs. Simple Structure):** Because PCA is bound to strict, orthogonal variance maximization, look at what happens to the remaining signals on `PC 2` and `PC 4`. `PC 2` and `PC 4` exhibit high loading values across *both* `Sensor_1` and `Sensor_2` simultaneously, while also bleeding into `Sensor_3`. This cross-loading represents a mathematically forced mixture of signals. Factor Analysis applies a **Varimax rotation** to eliminate this. By maximizing loading variance within columns, it completely isolates the signals: `Sensor_1` and `Sensor_2` group cleanly into `Factor 1`, while `Sensor_3` shifts entirely into `Factor 2`.
* **Part 2 (Operational Interpretability):** The PCA loading matrix forces a complex interpretation. If a diagnostic alarm trips on `PC 2`, an engineer has to figure out a messy combination of Sensors 1, 2, and 3. With the rotated FA framework, the layout is intuitive. `Factor 1` represents the isolated physical process causing redundancy in `Sensor_1` and `Sensor_2`, while `Factor 2` maps directly to the independent process monitored by `Sensor_3`. Troubleshooting root causes becomes instantaneous.


### Solution for Question 3: Determining Subspace Truncation ($k$)

* **Part 1 (Curve Observations):** Moving from $k=1$ to $k=2$, the Mean $Q$ Statistic (SPE) plunges from $2.4$ down to $0.9$. When transitioning from $k=2$ to $k=3$, the curve flattens out into an explicit "elbow", decreasing at a much slower, linear rate down toward $0.1$.
* **Part 2 (The True Intrinsic Dimension):** Cross-referencing the $Q$ curve with the new loading matrix makes the selection clear. `PC 1` tracks the isolated noise of `Sensor_4`. `PC 2` captures the true, shared variance of the working sensors (`Sensor_1`, `Sensor_2`, `Sensor_3`). The sharp drop in the $Q$ statistic at $k=2$ tells us that once these two distinct behaviors are accounted for, the structured information is fully exhausted. Thus, $k=2$ represents the true hidden physical dimensionality of the asset.
* **Part 3 (The Danger of Over-fitting to $k=3$):** Look at `PC 3` in the loading matrix—it maps exclusively to the leftover residual variance of `Sensor_3`. Setting your truncation limit to $k=3$ pulls this residual space into your "clean" model subspace. This overfits the system, causing your Hotelling's $T^2$ control limits to trigger frequent false alarms due to minor, normal baseline variations.



### Solution for Question 4: Operational Trade-offs in System Health Monitoring

* **The PCA Strategy:** Tracking the main principal subspace via $T^2$ alongside the residual space via the $Q$ statistic.
* *Vulnerability:* This approach is highly vulnerable to sensor degradation. As shown in the PCA loading matrix, if a single channel (`Sensor_4`) fails high, it structurally shifts the orientation of your principal axes, dragging the calculations for the remaining healthy channels along with it.


* **The FA Strategy:** Directly monitoring the rotated Latent Factor Scores.
* *Robustness:* This approach is highly robust. Because Factor Analysis explicitly separates localized noise via the diagonal uniqueness vector ($\Psi$), any single-channel sensor failure is safely absorbed by that channel's uniqueness profile ($\psi_4^2 \rightarrow 1$). The underlying latent factors (`Factor 1` and `Factor 2`) remain clean, stable, and uncorrupted, allowing the rest of your health monitoring network to function normally.
