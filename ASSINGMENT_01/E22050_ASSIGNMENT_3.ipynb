#Q1) Prediction of fraudulent transactions

A bank monitors credit-card transactions for possible fraud.

For each transaction, let the sample space be

$$
\Omega=\{\omega_1,\omega_2,\omega_3,\omega_4\},
$$

where:

* $\omega_1$: low-risk customer, legitimate transaction,
* $\omega_2$: low-risk customer, fraudulent transaction,
* $\omega_3$: high-risk customer, legitimate transaction,
* $\omega_4$: high-risk customer, fraudulent transaction.

Assume the probabilities are

$$
P(\{\omega_1\})=0.50,\qquad
P(\{\omega_2\})=0.10,\qquad
P(\{\omega_3\})=0.20,\qquad
P(\{\omega_4\})=0.20.
$$

Let $X:\Omega\to\mathbb R$ denote the fraud indicator:

$$
X(\omega_1)=0,\qquad
X(\omega_2)=1,\qquad
X(\omega_3)=0,\qquad
X(\omega_4)=1.
$$

Thus $X=1$ means the transaction is fraudulent, and $X=0$ means it is legitimate.

Now suppose the bank does not observe the full outcome $\omega$, but only the customer's risk class. Define $Y:\Omega\to\mathbb R$ by

$$
Y(\omega_1)=0,\qquad
Y(\omega_2)=0,\qquad
Y(\omega_3)=1,\qquad
Y(\omega_4)=1.
$$

Thus:

* $Y=0$ means “low-risk customer,”
* $Y=1$ means “high-risk customer.”

Answer the following.

1. Write down the probability space $(\Omega,\mathfrak F,P)$, where $\mathfrak F=\mathfrak P(\Omega)$.

2. Compute the $\sigma$-algebra generated by $Y$, namely $\sigma(Y)$.

3. Explain why $E[X\mid Y]$ must be $\sigma(Y)$-measurable, and therefore must be constant on the sets $\{ \omega_1,\omega_2\}$ and $\{\omega_3,\omega_4\}$.

4. Compute $E[X\mid Y]$ explicitly and show that

$$
E[X\mid Y]= \frac{1}{6} \mathbb{1}_{\{Y=0\}} + \frac{1}{2} \mathbb{1}_{\{Y=1\}}
$$

5. Verify the defining property of conditional expectation by checking that for every set $A\in\sigma(Y)$,

$$
\int_A E[X\mid Y]\,dP=\int_A X\,dP.
$$

6. Find the Conditional PMF $p_{X|Y}(x \mid y)$.

7. Verify the identity:
$$\mathbb{E}[X\mid Y=y] = \sum_{x \in \{0,1\}} x \cdot p_{X|Y}(x \mid y)$$.

8. Calculate the **Mean Square Error (MSE)**

9. How would the **Mean Square Error (MSE)** change if the bank decided to ignore the customer risk class $Y$ and instead used the global average fraud rate ($E[X] = 0.30$) as a constant estimate for every transaction? Calculate this error and explain its relationship to the **Law of Total Variance**.

10. Using the joint distribution of $(X, Y)$ derived from the sample space:

    a) Compute the entropy $H(X)$, representing the bank's initial uncertainty regarding fraud.

    b) Compute the conditional entropy $H(X \mid Y)$, representing the uncertainty remaining after the customer's risk class is revealed.

    c) Calculate the Mutual Information $I(X; Y)$.

    d) Based on these values, what percentage of the total uncertainty about fraud is removed by knowing the customer's risk class?

## Sample Answer

### Q1.1

The probability space is

$$ \Omega=\{\omega_1,\omega_2,\omega_3,\omega_4\}, $$

$$ \mathfrak F=\mathfrak P(\Omega), $$

and the probability measure $P$ is defined by

$$ P(\{\omega_1\})=0.50,\qquad P(\{\omega_2\})=0.10,\qquad P(\{\omega_3\})=0.20,\qquad P(\{\omega_4\})=0.20. $$

Since

$$ 0.50+0.10+0.20+0.20=1, $$

this is a valid probability measure.

---

### Verification Conditions:

1. **Non-negativity:**
   $$ P(\{\omega_i\}) \ge 0 \quad \text{for all } i \in \{1, 2, 3, 4\} $$

2. **Normalization:**
   $$ \sum_{i=1}^{4} P(\{\omega_i\}) = 0.50 + 0.10 + 0.20 + 0.20 = 1 $$

###Q1.2

The random variable $Y$ is defined by

$$ Y(\omega_1)=0,\qquad Y(\omega_2)=0,\qquad Y(\omega_3)=1,\qquad Y(\omega_4)=1. $$

Thus the inverse images of the values of $Y$ are

$$ Y^{-1}(\{0\})=\{\omega_1,\omega_2\}, $$

and

$$ Y^{-1}(\{1\})=\{\omega_3,\omega_4\}. $$

Therefore the $\sigma$-algebra generated by $Y$ is

$$ \sigma(Y)=\left\{\varnothing,\Omega,\{\omega_1,\omega_2\},\{\omega_3,\omega_4\}\right\}. $$

---

### Verification:
A collection of subsets $\mathcal{A}$ is a $\sigma$-algebra if it satisfies three properties:
1. **Contains the empty set:** $\varnothing \in \sigma(Y)$
2. **Closed under complementation:**
   * $\varnothing^c = \Omega \in \sigma(Y)$
   * $\{\omega_1, \omega_2\}^c = \{\omega_3, \omega_4\} \in \sigma(Y)$
3. **Closed under countable unions:**
   * $\{\omega_1, \omega_2\} \cup \{\omega_3, \omega_4\} = \Omega \in \sigma(Y)$

### Q1.3

By definition, $E[X\mid Y]$ must be $\sigma(Y)$-measurable. This means it can depend only on the value of $Y$, not on the full outcome $\omega$. That is, $E[X\mid Y]=g(Y)$ for some measurable map $g$.

Since $\sigma(Y)$ can distinguish only the two sets

$$ \{\omega_1,\omega_2\} \quad \text{and} \quad \{\omega_3,\omega_4\}, $$

the random variable $E[X\mid Y]$ must be constant on each of these sets.

So we write

$$
E[X\mid Y](\omega)=
\begin{cases}
a, & \omega\in\{\omega_1,\omega_2\},\\
b, & \omega\in\{\omega_3,\omega_4\},
\end{cases}
$$

for some constants $a$ and $b$ to be determined.

---

### Definition Property for Calculations:
To find $a$ and $b$, use the partial averaging property for each atom $A \in \sigma(Y)$:

$$ \sum_{\omega \in A} E[X \mid Y](\omega) P(\{\omega\}) = \sum_{\omega \in A} X(\omega) P(\{\omega\}) $$

* **For $A = \{\omega_1, \omega_2\}$ (where $Y=0$):**
  $$ a \cdot P(\{\omega_1\}) + a \cdot P(\{\omega_2\}) = X(\omega_1)P(\{\omega_1\}) + X(\omega_2)P(\{\omega_2\}) $$

* **For $A = \{\omega_3, \omega_4\}$ (where $Y=1$):**
  $$ b \cdot P(\{\omega_3\}) + b \cdot P(\{\omega_4\}) = X(\omega_3)P(\{\omega_3\}) + X(\omega_4)P(\{\omega_4\}) $$

###Q1.4

To determine the constants $a$ and $b$, we apply the defining characteristic of conditional expectation across the partitions of our space.

First, evaluating over the subset $\{\omega_1,\omega_2\}$:

$$ \int_{\{\omega_1,\omega_2\}} E[X\mid Y]\,dP = \int_{\{\omega_1,\omega_2\}} X\,dP $$

Expanding the expectation side yields:

$$ a \cdot P(\{\omega_1,\omega_2\}) = a(0.50 + 0.10) = 0.60a $$

Computing the actual expectation of $X$ on this subset gives:

$$ 0 \cdot 0.50 + 1 \cdot 0.10 = 0.10 $$

Equating the two sides:

$$ 0.60a = 0.10 $$

$$ a = \frac{0.10}{0.60} = \frac{1}{6} $$

---

Next, evaluating over the remaining subset $\{\omega_3,\omega_4\}$:

$$ \int_{\{\omega_3,\omega_4\}} E[X\mid Y]\,dP = \int_{\{\omega_3,\omega_4\}} X\,dP $$

Expanding the expectation side yields:

$$ b \cdot P(\{\omega_3,\omega_4\}) = b(0.20 + 0.20) = 0.40b $$

Computing the actual expectation of $X$ on this subset gives:

$$ 0 \cdot 0.20 + 1 \cdot 0.20 = 0.20 $$

Equating the two sides:

$$ 0.40b = 0.20 $$

$$ b = \frac{0.20}{0.40} = \frac{1}{2} $$

---

Consequently, the pointwise conditional expectation mapping is defined as:

$$
E[X\mid Y](\omega) =
\begin{cases}
\dfrac{1}{6}, & \omega\in\{\omega_1,\omega_2\},\\
\dfrac{1}{2}, & \omega\in\{\omega_3,\omega_4\}.
\end{cases}
$$

Represented more compactly as a random variable dependent on $Y$ via indicator functions:

$$
E[X\mid Y] = \frac{1}{6} \mathbb{1}_{\{Y=0\}} + \frac{1}{2} \mathbb{1}_{\{Y=1\}} =
\begin{cases}
\dfrac{1}{6}, & Y=0,\\
\dfrac{1}{2}, & Y=1.
\end{cases}
$$

###Q1.5

We will now validate the foundational conditional expectation identity:

$$ \int_A E[X\mid Y]\,dP = \int_A X\,dP \qquad \forall A\in\sigma(Y). $$

The collection of target events within $\sigma(Y)$ consists of:

$$ \varnothing,\qquad \Omega,\qquad \{\omega_1,\omega_2\},\qquad \{\omega_3,\omega_4\}. $$

---

### Verifying Every Sub-Event Case:

* **Case 1: $A = \varnothing$**  
  Trivially, integrating over an empty domain yields:
  $$ \int_{\varnothing} E[X\mid Y]\,dP = \int_{\varnothing} X\,dP = 0 $$

* **Case 2: $A = \{\omega_1,\omega_2\}$**  
  Evaluating the left-hand integral using our determined value:
  $$ \int_{\{\omega_1,\omega_2\}} E[X\mid Y]\,dP = \frac{1}{6} \cdot P(\{\omega_1,\omega_2\}) = \frac{1}{6} \cdot 0.60 = 0.10 $$
  Evaluating the true variable profile on the right-hand side:
  $$ \int_{\{\omega_1,\omega_2\}} X\,dP = X(\omega_1)P(\{\omega_1\}) + X(\omega_2)P(\{\omega_2\}) = 0 \cdot 0.50 + 1 \cdot 0.10 = 0.10 $$

* **Case 3: $A = \{\omega_3,\omega_4\}$**  
  Evaluating the left-hand integral:
  $$ \int_{\{\omega_3,\omega_4\}} E[X\mid Y]\,dP = \frac{1}{2} \cdot P(\{\omega_3,\omega_4\}) = \frac{1}{2} \cdot 0.40 = 0.20 $$
  Evaluating the right-hand integral:
  $$ \int_{\{\omega_3,\omega_4\}} X\,dP = X(\omega_3)P(\{\omega_3\}) + X(\omega_4)P(\{\omega_4\}) = 0 \cdot 0.20 + 1 \cdot 0.20 = 0.20 $$

* **Case 4: $A = \Omega$**  
  Summing across the full partition for the conditional estimator:
  $$ \int_\Omega E[X\mid Y]\,dP = \left(\frac{1}{6} \cdot 0.60\right) + \left(\frac{1}{2} \cdot 0.40\right) = 0.10 + 0.20 = 0.30 $$
  Summing across the entire sample space for the raw variable $X$:
  $$ \int_\Omega X\,dP = (0 \cdot 0.50) + (1 \cdot 0.10) + (0 \cdot 0.20) + (1 \cdot 0.20) = 0.30 $$

---

### Analytical Conclusion:
The core metric balances identically for all $A \in \sigma(Y)$.

Crucially, the final step across the full domain $\Omega$ serves as a direct, concrete proof of the **Law of Total Expectation**:

$$ E[E[X\mid Y]] = E[X] = 0.30 $$

###Q1.6

Applying the definition of conditional probability via the product rule $p_{X|Y}(x \mid y) = \frac{P(X=x, Y=y)}{P(Y=y)}$, we isolate the conditional likelihood of a fraudulent event ($X=1$) across both distinct risk classifications:

* **Evaluating the Low-Risk Category ($Y=0$):**
  $$ p_{X|Y}(1 \mid 0) = \frac{P(X=1, \cap \, Y=0)}{P(Y=0)} = \frac{P(\{\omega_2\})}{0.60} = \frac{0.10}{0.60} = \frac{1}{6} $$

* **Evaluating the High-Risk Category ($Y=1$):**
  $$ p_{X|Y}(1 \mid 1) = \frac{P(X=1, \cap \, Y=1)}{P(Y=1)} = \frac{P(\{\omega_4\})}{0.40} = \frac{0.20}{0.40} = \frac{1}{2} $$

*(Note: The corresponding non-fraud probabilities ($X=0$) can be derived directly using complements, yielding $5/6$ for low-risk and $1/2$ for high-risk conditions.)*

###Q1.8

We evaluate the Minimum Mean Square Error (MMSE), which identifies the optimal irreducible uncertainty remaining when we apply the regression function $\mathbb{E}[X|Y]$ as our predictor. As established by the orthogonality principle, this minimum error equals the expected conditional variance:

$$ \text{MSE}_{\text{min}} = \mathbb{E}\left[(X - \mathbb{E}[X \mid Y])^2\right] = \mathbb{E}[\mathrm{Var}(X \mid Y)] $$

---

### Phase 1: Isolating Group Variances
We first calculate the conditional variance profile for each structural subgroup $y \in \{0, 1\}$. Because the conditional distribution of our indicator $X$ given $Y$ matches a Bernoulli distribution, the conditional variance simplifies directly to $p(1-p)$:

* **For the Baseline Tier ($Y=0$):**
  $$ \mathrm{Var}(X \mid Y=0) = \frac{1}{6} \left(1 - \frac{1}{6}\right) = \frac{1}{6} \cdot \frac{5}{6} = \frac{5}{36} $$

* **For the Elevated Tier ($Y=1$):**
  $$ \mathrm{Var}(X \mid Y=1) = \frac{1}{2} \left(1 - \frac{1}{2}\right) = \frac{1}{2} \cdot \frac{1}{2} = \frac{1}{4} $$

---

### Phase 2: Integrating Over the Marginal Probability of $Y$
Next, we weight these conditional variances against the prior probabilities of the grouping variable, where $P(Y=0) = 0.60 = \frac{3}{5}$ and $P(Y=1) = 0.40 = \frac{2}{5}$:

$$ \mathbb{E}[\mathrm{Var}(X \mid Y)] = \mathrm{Var}(X \mid Y=0)P(Y=0) + \mathrm{Var}(X \mid Y=1)P(Y=1) $$

Substituting our fractional values into the expectation equation:

$$ \mathbb{E}[\mathrm{Var}(X \mid Y)] = \left(\frac{5}{36} \cdot \frac{3}{5}\right) + \left(\frac{1}{4} \cdot \frac{2}{5}\right) $$

$$ \mathbb{E}[\mathrm{Var}(X \mid Y)] = \frac{1}{12} + \frac{1}{10} = \frac{5+6}{60} = \frac{11}{60} \approx 0.1833 $$

---

### Statistical Significance:
The resulting value of $\frac{11}{60} \approx 0.1833$ defines the absolute floor for the mean squared error achievable under this information structure. Any alternative estimator or heuristic utilizing these specific classification criteria will inevitably generate a total squared error that exceeds this fundamental limit.

The **Law of Total Variance** partitions total global uncertainty into two explicit mathematical categories:

Mapping these variables to the operational metrics:

* **$\mathrm{Var}(\mathbb{E}[X \mid Y])$ (Explained Variance / Error Reduction):** This component measures the exact drop in system uncertainty generated by conditioning on $Y$.

  $$ \mathrm{Var}(\mathbb{E}[X \mid Y]) = \mathbb{E}\left[(\mathbb{E}[X \mid Y])^2\right] - (\mathbb{E}[X])^2 $$

  $$ \mathrm{Var}(\mathbb{E}[X \mid Y]) = \left[\frac{1}{60} + \frac{1}{10}\right] - 0.09 = \frac{7}{60} - \frac{5.4}{60} = \frac{1.6}{60} \approx 0.0267 $$

---

### Final Blueprint Summary

Aggregating the variance parts provides an exact structural match:

$$ 0.2100 \, (\text{Total System Error}) = 0.1833 \, (\text{Irreducible Noise}) + 0.0267 \, (\text{Error Reduction}) $$

The Law of Total Variance provides the ultimate mathematical guarantee that an informed conditioning strategy will always minimize or match the global baseline error. The difference between the two configurations is exactly the variance of the conditional estimator, isolating the practical value that feature $Y$ introduces into the system.

# Q2) Conditional Density of the Multivariate Gaussian

Let $\mathbf{Z}$ be the concatenated random matrix
$\mathbf{Z} = \begin{bmatrix} \mathbf{X} \\ \mathbf{Y} \end{bmatrix} \in \mathbb{R}^{n+m}$.


The random variable $\mathbf{Z}$ is said to have a **joint multivariate normal distribution** if every linear combination of its components follows a univariate normal distribution.

For $\mathbf{Z}$ to possess a **joint density function** (meaning its distribution is absolutely continuous with respect to the Lebesgue measure on $\mathbb{R}^{n+m}$), its covariance matrix must be non-singular (positive definite).


The joint density function $f_{\mathbf{X}, \mathbf{Y}}(\mathbf{z})$ for $\mathbf{z} \in \mathbb{R}^{n+m}$ is given by:

$$f_{\mathbf{X}, \mathbf{Y}}(\mathbf{z}) = \frac{1}{\sqrt{(2\pi)^{n+m} |\boldsymbol{\Sigma}|}} \exp\left( -\frac{1}{2} (\mathbf{z} - \boldsymbol{\mu})^T \boldsymbol{\Sigma}^{-1} (\mathbf{z} - \boldsymbol{\mu}) \right).$$

Here:

* **Mean ($\boldsymbol{\mu}$):** $\boldsymbol{\mu} = E[\mathbf{Z}] = \begin{bmatrix} \boldsymbol{\mu}_X \\ \boldsymbol{\mu}_Y \end{bmatrix} \in \mathbb{R}^{n+m}$, where $\boldsymbol{\mu}_X \in \mathbb{R}^n$ and $\boldsymbol{\mu}_Y \in \mathbb{R}^m$.
* **Covariance Matrix ($\boldsymbol{\Sigma}$):** A symmetric, positive definite matrix $\boldsymbol{\Sigma} \in \mathbb{R}^{(n+m) \times (n+m)}$ partitioned as:
$$\boldsymbol{\Sigma} = \begin{bmatrix} \boldsymbol{\Sigma}_{XX} & \boldsymbol{\Sigma}_{XY} \\ \boldsymbol{\Sigma}_{YX} & \boldsymbol{\Sigma}_{YY} \end{bmatrix}$$
    * $\boldsymbol{\Sigma}_{XX} = \text{Var}(\mathbf{X})$ (size $n \times n$)
    * $\boldsymbol{\Sigma}_{YY} = \text{Var}(\mathbf{Y})$ (size $m \times m$)
    * $\boldsymbol{\Sigma}_{XY} = \text{Cov}(\mathbf{X}, \mathbf{Y}) = \boldsymbol{\Sigma}_{YX}^T$ (size $n \times m$)
* **Determinant ($|\boldsymbol{\Sigma}|$):** The normalization factor ensuring the total integral is 1.

Show that the conditional distribution is given by $f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x} \mid \mathbf{y})$

$$f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x} \mid \mathbf{y}) = \frac{1}{\sqrt{(2\pi)^n |\mathbf{S}_{XX}|}} \exp\left( -\frac{1}{2} (\mathbf{x} - \boldsymbol{\mu}^*)^T \boldsymbol{\Sigma}_{X\mid Y}^{-1} (\mathbf{x} - \boldsymbol{\mu}^*) \right),$$
where
* $\boldsymbol{\Sigma}_{X\mid Y} = \boldsymbol{\Sigma}_{XX} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX}$
* $\boldsymbol{\mu}^* = \boldsymbol{\mu}_x + \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}(\mathbf{y} - \boldsymbol{\mu}_y)$

## Sample Answer

To derive the conditional density function $f_{\mathbf{X}\mid\mathbf{Y}}(\mathbf{x} \mid \mathbf{y})$, we evaluate the continuous probability quotient:

$$ f_{\mathbf{X}\mid\mathbf{Y}}(\mathbf{x} \mid \mathbf{y}) = \frac{f_{\mathbf{X}, \mathbf{Y}}(\mathbf{x}, \mathbf{y})}{f_{\mathbf{Y}}(\mathbf{y})} $$

The derivation utilizes block matrix factorization via the Schur complement to decouple the joint quadratic form and decompose the determinant.

---

### Phase 1: Determinant Factorization via Schur Complement

Let $\boldsymbol{\Sigma}_{X\mid Y} = \boldsymbol{\Sigma}_{XX} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX}$ define the Schur complement of $\boldsymbol{\Sigma}_{YY}$ in the joint covariance matrix $\boldsymbol{\Sigma}$. We unpack the partitioned covariance structure using the following exact block matrix decomposition:

$$ \boldsymbol{\Sigma} = \begin{bmatrix} \mathbf{I}_n & \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1} \\ \mathbf{0} & \mathbf{I}_m \end{bmatrix} \begin{bmatrix} \boldsymbol{\Sigma}_{X\mid Y} & \mathbf{0} \\ \mathbf{0} & \boldsymbol{\Sigma}_{YY} \end{bmatrix} \begin{bmatrix} \mathbf{I}_n & \mathbf{0} \\ \boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX} & \mathbf{I}_m \end{bmatrix} $$

Taking the determinant of this product isolates the scaling factors because the upper and lower triangular block matrices have determinants equal to 1:

$$ |\boldsymbol{\Sigma}| = |\boldsymbol{\Sigma}_{X\mid Y}| \cdot |\boldsymbol{\Sigma}_{YY}| $$

---

### Phase 2: Inverting the Partitioned Covariance Matrix

By applying matrix inversion rules to our block factorization, we obtain a precise representation for the inverse covariance operator $\boldsymbol{\Sigma}^{-1}$:

$$ \boldsymbol{\Sigma}^{-1} = \begin{bmatrix} \mathbf{I}_n & \mathbf{0} \\ -\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX} & \mathbf{I}_m \end{bmatrix} \begin{bmatrix} \boldsymbol{\Sigma}_{X\mid Y}^{-1} & \mathbf{0} \\ \mathbf{0} & \boldsymbol{\Sigma}_{YY}^{-1} \end{bmatrix} \begin{bmatrix} \mathbf{I}_n & -\boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1} \\ \mathbf{0} & \mathbf{I}_m \end{bmatrix} $$

---

### Phase 3: Decomposing the Joint Quadratic Form

Let $\tilde{\mathbf{x}} = \mathbf{x} - \boldsymbol{\mu}_X$ and $\tilde{\mathbf{y}} = \mathbf{y} - \boldsymbol{\mu}_Y$. We inject our factored expression for $\boldsymbol{\Sigma}^{-1}$ directly into the core exponential argument $(\mathbf{z} - \boldsymbol{\mu})^T \boldsymbol{\Sigma}^{-1} (\mathbf{z} - \boldsymbol{\mu})$:

$$ (\mathbf{z} - \boldsymbol{\mu})^T \boldsymbol{\Sigma}^{-1} (\mathbf{z} - \boldsymbol{\mu}) = \begin{bmatrix} \tilde{\mathbf{x}} \\ \tilde{\mathbf{y}} \end{bmatrix}^T \begin{bmatrix} \mathbf{I}_n & \mathbf{0} \\ -\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX} & \mathbf{I}_m \end{bmatrix} \begin{bmatrix} \boldsymbol{\Sigma}_{X\mid Y}^{-1} & \mathbf{0} \\ \mathbf{0} & \boldsymbol{\Sigma}_{YY}^{-1} \end{bmatrix} \begin{bmatrix} \mathbf{I}_n & -\boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1} \\ \mathbf{0} & \mathbf{I}_m \end{bmatrix} \begin{bmatrix} \tilde{\mathbf{x}} \\ \tilde{\mathbf{y}} \end{bmatrix} $$

We simplify the rightmost vector multiplication first:

$$ \begin{bmatrix} \mathbf{I}_n & -\boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1} \\ \mathbf{0} & \mathbf{I}_m \end{bmatrix} \begin{bmatrix} \tilde{\mathbf{x}} \\ \tilde{\mathbf{y}} \end{bmatrix} = \begin{bmatrix} \tilde{\mathbf{x}} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\tilde{\mathbf{y}} \\ \tilde{\mathbf{y}} \end{bmatrix} $$

Notice that the top block exactly matches the shifted conditional mean profile $\mathbf{x} - \boldsymbol{\mu}^*$:

$$ \tilde{\mathbf{x}} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\tilde{\mathbf{y}} = (\mathbf{x} - \boldsymbol{\mu}_X) - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}(\mathbf{y} - \boldsymbol{\mu}_Y) = \mathbf{x} - \boldsymbol{\mu}^* $$

Substituting this transformation back into the full quadratic expression decouples the joint vector into two separate symmetric forms:

$$ (\mathbf{z} - \boldsymbol{\mu})^T \boldsymbol{\Sigma}^{-1} (\mathbf{z} - \boldsymbol{\mu}) = (\mathbf{x} - \boldsymbol{\mu}^*)^T \boldsymbol{\Sigma}_{X\mid Y}^{-1} (\mathbf{x} - \boldsymbol{\mu}^*) + \tilde{\mathbf{y}}^T \boldsymbol{\Sigma}_{YY}^{-1} \tilde{\mathbf{y}} $$

---

### Phase 4: Constructing the Probability Density Quotient

We now substitute the split quadratic exponent and the factored determinant term back into the numerator of our conditional definition:

$$ f_{\mathbf{X}, \mathbf{Y}}(\mathbf{x}, \mathbf{y}) = \frac{1}{\sqrt{(2\pi)^{n+m} |\boldsymbol{\Sigma}_{X\mid Y}| \cdot |\boldsymbol{\Sigma}_{YY}|}} \exp\left( -\frac{1}{2} \left[ (\mathbf{x} - \boldsymbol{\mu}^*)^T \boldsymbol{\Sigma}_{X\mid Y}^{-1} (\mathbf{x} - \boldsymbol{\mu}^*) + (\mathbf{y} - \boldsymbol{\mu}_Y)^T \boldsymbol{\Sigma}_{YY}^{-1} (\mathbf{y} - \boldsymbol{\mu}_Y) \right] \right) $$

Factoring this product separates the expressions completely:

$$ f_{\mathbf{X}, \mathbf{Y}}(\mathbf{x}, \mathbf{y}) = \left[ \frac{1}{\sqrt{(2\pi)^n |\boldsymbol{\Sigma}_{X\mid Y}|}} \exp\left( -\frac{1}{2}(\mathbf{x} - \boldsymbol{\mu}^*)^T \boldsymbol{\Sigma}_{X\mid Y}^{-1} (\mathbf{x} - \boldsymbol{\mu}^*) \right) \right] \times \left[ \frac{1}{\sqrt{(2\pi)^m |\boldsymbol{\Sigma}_{YY}|}} \exp\left( -\frac{1}{2}(\mathbf{y} - \boldsymbol{\mu}_Y)^T \boldsymbol{\Sigma}_{YY}^{-1} (\mathbf{y} - \boldsymbol{\mu}_Y) \right) \right] $$

The second bracketed term is identical to the marginal multivariate normal distribution density $f_{\mathbf{Y}}(\mathbf{y})$. Dividing the joint distribution by this marginal density eliminates it:

$$ f_{\mathbf{X}\mid\mathbf{Y}}(\mathbf{x} \mid \mathbf{y}) = \frac{f_{\mathbf{X}, \mathbf{Y}}(\mathbf{x}, \mathbf{y})}{f_{\mathbf{Y}}(\mathbf{y})} = \frac{1}{\sqrt{(2\pi)^n |\boldsymbol{\Sigma}_{X\mid Y}|}} \exp\left( -\frac{1}{2} (\mathbf{x} - \boldsymbol{\mu}^*)^T \boldsymbol{\Sigma}_{X\mid Y}^{-1} (\mathbf{x} - \boldsymbol{\mu}^*) \right) $$

This matches the target structure exactly, showing that the conditional distribution is a multivariate normal distribution with mean $\boldsymbol{\mu}^*$ and covariance matrix $\boldsymbol{\Sigma}_{X\mid Y}$ (which corresponds to $\mathbf{S}_{XX}$).

# Q3) Continuous Fault Monitoring of a Spindle Motor

A factory monitors the health of a high-speed CNC spindle. The critical variable is the **Internal Temperature ($X$)**, but because the motor is sealed, they can only observe the **Total Spindle Power ($Y$)**.

**The Setup:**

    $$X \sim \mathscr{N}(0, 1)$$
2.  **Mechanical Load ($Q$):** The torque required to cut material creates "noise" in the power reading. It is independent of temperature and follows:
    $$Q \sim \mathscr{N}(0, 4)$$
3.  **The Measurement ($Y$):** The power sensor $Y$ measures the combined effect of thermal load and mechanical torque:
$$Y = X + Q$$

---

**Answer the followings:**

**1. Joint Distribution**
Identify the distribution of the random varaible $(X, Y)$.
* Show that $Y \sim \mathscr{N}(0, 5)$.
* Find the covariance $\text{Cov}(X, Y)=\Sigma_{XY}$.

**2. Information Geometry**
Explain why $X$ is **not** $\sigma(Y)$-measurable. In practical terms, if the power meter shows $Y = 5$, why can't the engineer know for certain if the motor is overheating ($X$ is high) or if the material being cut is simply very hard ($Q$ is high)?

**3. Conditional Expectation (The Best Estimate)**
Compute the conditional expectation $E[X \mid Y]$.
* Using the property for joint normals: $E[X \mid Y = y] = \mu_X + \frac{\text{Cov}(X,Y)}{\text{Var}(Y)}(y - \mu_Y)$.
* Express this as a "signal-to-total-variance" ratio.

**4. Verification of the Residual**
Let $\widehat{X} = E[X \mid Y]$. Define the estimation error (residual) as $Z = X - \widehat{X}$.
* Prove that $E[Z] = 0$.
* Verify that the error $Z$ is uncorrelated with the observation $Y$ (the Orthogonality Principle).

**5. Information Theory: Differential Entropy**
* Compute the differential entropy $h(X)$.
* Compute the conditional differential entropy $h(X \mid Y)$.
* *Note: The variance of the conditional distribution $(X \mid Y=y)$ is given by $\sigma_{X|Y}^2 = \sigma_X^2 - \frac{\text{Cov}(X,Y)^2}{\text{Var}(Y)}$.*

**6. Mutual Information**
Calculate $I(X; Y) = h(X) - h(X \mid Y)$.
* What percentage of the "uncertainty" about the temperature is removed by observing the power meter?
* How would this change if the mechanical noise $Q$ had a variance of $100$ instead of $4$?

**7. Diagnostic Interpretation**
The factory manager says: *"Since $Y = X + Q$, our best estimate of temperature should just be $Y$ itself."*
Using your result from Question 3, explain why $Y$ is a **biased** estimator of $X$ and why $E[X \mid Y]$ "shrinks" the measurement toward the mean (zero).



## Sample Answer

To prepare these statistical and information-theoretic derivations for clear rendering in your Google Colab notebook, we map out the joint space and the corresponding signal-processing mechanics below:

---

### 1. Joint Distribution Architecture of $(X, Y)$

Given that the core thermal parameter $X$ and the operational noise $Q$ act as independent Gaussian variables ($X \perp Q$), their linear combination $Y = X + Q$ preserves a normal distribution profile:

* **Expected Value:**
  $$ \mathbb{E}[Y] = \mathbb{E}[X] + \mathbb{E}[Q] = 0 + 0 = 0 $$
* **Total Variance:** Because the variables are completely orthogonal, the independent variances add up directly:
  $$ \mathrm{Var}(Y) = \mathrm{Var}(X) + \mathrm{Var}(Q) = 1 + 4 = 5 $$
* **Cross-Covariance Form:** We resolve the interaction between the system parameter and the composite signal as follows:
  $$ \mathrm{Cov}(X, Y) = \mathbb{E}[XY] - \mathbb{E}[X]\mathbb{E}[Y] = \mathbb{E}[X(X+Q)] = \mathbb{E}[X^2] + \mathbb{E}[XQ] $$
  Substituting $\mathbb{E}[X^2] = \mathrm{Var}(X) = 1$ and $\mathbb{E}[XQ] = \mathbb{E}[X]\mathbb{E}[Q] = 0$ yields:
  $$ \mathrm{Cov}(X, Y) = 1 $$

Expressing this system structure in standard joint multivariate normal format:

$$ \begin{bmatrix} X \\ Y \end{bmatrix} \sim \mathcal{N} \left( \begin{bmatrix} 0 \\ 0 \end{bmatrix}, \begin{bmatrix} 1 & 1 \\ 1 & 5 \end{bmatrix} \right) $$

---

### 2. Geometric Measurability Constraint

The underlying signal $X$ is not strictly $\sigma(Y)$-measurable because there is no unique deterministic inversion mapping $g$ such that $X = g(Y)$. For any observed metric level $Y=y$, an infinite set of candidate coordinate pairs $(x, q)$ satisfies the constraint $x + q = y$.

> **Physical Implication:** An isolated elevated sensor reading $Y$ remains structurally ambiguous. It can reflect a genuine internal thermal anomaly (high $X$) or a dense mechanical load interaction (high $Q$). The monitoring channel $Y$ cannot decouple these distinct states on its own.

---

### 3. Resolving the Optimal Conditional Estimator

Under joint Gaussian structures, the minimum mean-square error (MMSE) prediction is an explicitly linear function:

$$ \mathbb{E}[X \mid Y = y] = \mu_X + \frac{\mathrm{Cov}(X,Y)}{\mathrm{Var}(Y)}(y - \mu_Y) = 0 + \frac{1}{5}(y - 0) $$

$$ \mathbb{E}[X \mid Y] = 0.2 Y $$

> **Operational Weighting:** Because the external environmental noise ($Q$) exhibits four times the variance of the true signal ($X$), the optimization logic scales down the reading, trusting exactly 20% of the overall observed volatility as thermal variance.

---

### 4. Verification via Orthogonal Residuals

Let $\widehat{X} = 0.2 Y$ serve as our estimator. We track the behavior of the tracking error $Z = X - 0.2 Y$:

* **Residual Mean:**  
  $$ \mathbb{E}[Z] = \mathbb{E}[X] - 0.2 \mathbb{E}[Y] = 0 - 0 = 0 $$
* **Orthogonality Condition:** We check the covariance between the error term and the observation space to ensure no data leaks:
  $$ \mathrm{Cov}(Z, Y) = \mathbb{E}[ZY] = \mathbb{E}[(X - 0.2Y)Y] = \mathbb{E}[XY] - 0.2 \mathbb{E}[Y^2] $$
  Substituting the known raw moments $\mathbb{E}[XY] = 1$ and $\mathbb{E}[Y^2] = 5$:
  $$ \mathrm{Cov}(Z, Y) = 1 - 0.2(5) = 1 - 1 = 0 $$

The tracking error remains completely uncorrelated with our observation channel, confirming that all accessible information has been extracted.

---

### 5. Continuous Differential Entropy Profiles

Using the continuous Gaussian definition $h(\mathcal{N}(\mu, \sigma^2)) = \frac{1}{2}\ln(2\pi e \sigma^2)$, we isolate the informational entropy metrics:

* **Unconditional Target Entropy $h(X)$:** Evaluating at baseline variance $\sigma_X^2 = 1$:
  $$ h(X) = \frac{1}{2}\ln(2\pi e) \approx 1.4189 \text{ nats} $$
* **Conditional Residual Entropy $h(X \mid Y)$:** We first compute the remaining conditional variance:
  $$ \sigma_{X\mid Y}^2 = \sigma_X^2 - \frac{\mathrm{Cov}(X,Y)^2}{\mathrm{Var}(Y)} = 1 - \frac{1}{5} = 0.8 $$
  $$ h(X \mid Y) = \frac{1}{2}\ln(2\pi e \cdot 0.8) \approx 1.3073 \text{ nats} $$

---

### 6. Mutual Information Metric

The net reduction in target system uncertainty is determined via entropy divergence:

$$ I(X; Y) = h(X) - h(X \mid Y) = \frac{1}{2}\ln(1) - \frac{1}{2}\ln(0.8) = \frac{1}{2}\ln\left(\frac{1}{0.8}\right) = \frac{1}{2}\ln(1.25) \approx 0.1116 \text{ nats} $$

> **Uncertainty Reduction:** This optimization drops the predictive variance from $1.0$ down to $0.8$, yielding a precise **20% improvement** in tracking resolution. If environmental noise rises sharply (e.g., $\sigma_Q^2 = 100$), the mutual information $I(X;Y)$ drops toward zero, meaning the signal gets completely overwhelmed by background noise.

---

### 7. Comparative Diagnostic Advantage

Relying on a direct raw matching strategy $X = Y$ represents a **naive baseline configuration**.
* Under the naive assumption, the resulting error profile corresponds directly to the noise term: $\mathbb{E}[(X-Y)^2] = \mathbb{E}[Q^2] = 4$.
* Implementing the structured MMSE configuration $\mathbb{E}[X \mid Y] = 0.2Y$ limits the tracking error variance to exactly $0.8$.

**Shrinkage Mechanism:** The optimal parameter $0.2$ serves as a statistical shrinkage coefficient. When data collection channels are compromised by high volatility, the ideal response is to pull the raw sensor readings back toward the historical baseline mean. If a sensor value spikes, the model avoids assuming a terminal thermal fault; instead, it correctly attributes the bulk of that deviation to mechanical load noise.

```python
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots
from scipy.stats import multivariate_normal, norm

# --- 1. Parameters for the Spindle Problem ---
# X: Temperature ~ N(0, 1)
# Q: Torque ~ N(0, 4)
# Y: Power = X + Q => Y ~ N(0, 5)
mu_x, mu_y = 0, 0
var_x, var_q = 1, 4
var_y = var_x + var_q
cov_xy = var_x  # Cov(X, X+Q) = Var(X) = 1
rho = cov_xy / (np.sqrt(var_x) * np.sqrt(var_y))

# Shrinkage Factor (Gain)
k = cov_xy / var_y  # 1/5 = 0.2

# --- 2. Generate Data for Visualization ---
# Sample points to show the "Cloud" of data
np.random.seed(42)
n_samples = 1000
x_samples = np.random.normal(mu_x, np.sqrt(var_x), n_samples)
q_samples = np.random.normal(0, np.sqrt(var_q), n_samples)
y_samples = x_samples + q_samples
x_hat = k * y_samples  # E[X|Y]
residual = x_samples - x_hat

# Grids for Contours
x_grid = np.linspace(-4, 4, 200)
y_grid = np.linspace(-8, 8, 200)
X_m, Y_m = np.meshgrid(x_grid, y_grid)
pos = np.dstack((X_m, Y_m))
Z = multivariate_normal([mu_x, mu_y], [[var_x, cov_xy], [cov_xy, var_y]]).pdf(pos)

# --- 3. Create Simulation Plot ---
fig = make_subplots(
    rows=1, cols=2,
    subplot_titles=(
        "Joint Distribution: X vs Y<br><sup>The Line of Best Estimate (Shrinkage)</sup>",
        "Estimation Error (Residual) vs Y<br><sup>Visualizing the Orthogonality Principle</sup>"
    )
)

# --- LEFT PLOT: Joint X, Y and the Estimator ---
# Density Contour
fig.add_trace(go.Contour(
    x=x_grid, y=y_grid, z=Z, colorscale='Blues', showscale=False, opacity=0.6,
    hoverinfo='skip'
), row=1, col=1)

# Scatter of actual realizations
fig.add_trace(go.Scatter(
    x=x_samples, y=y_samples, mode='markers',
    marker=dict(color='rgba(0,0,0,0.1)', size=3),
    name="Realized (x, y)"
), row=1, col=1)

# The Best Estimate Line E[X|Y] = 0.2Y
y_line = np.linspace(-8, 8, 10)
fig.add_trace(go.Scatter(
    x=k * y_line, y=y_line,
    line=dict(color='Red', width=4),
    name="E[X|Y] = 0.2Y (Optimal)"
), row=1, col=1)

# The Naive Estimator Line X = Y
fig.add_trace(go.Scatter(
    x=y_line, y=y_line,
    line=dict(color='Black', width=2, dash='dash'),
    name="X = Y (Naive)"
), row=1, col=1)

# --- RIGHT PLOT: Orthogonality (Residual vs Y) ---
fig.add_trace(go.Scatter(
    x=y_samples, y=residual, mode='markers',
    marker=dict(color='Purple', opacity=0.4, size=4),
    name="Residual (X - 0.2Y)"
), row=1, col=2)

# Horizontal line at 0 to show no correlation
fig.add_trace(go.Scatter(
    x=y_line, y=np.zeros_like(y_line),
    line=dict(color='Black', width=2),
    name="Zero Mean Error"
), row=1, col=2)

# --- Layout and Styling ---
fig.update_layout(
    height=600,
    title_text="Spindle Motor Diagnostic Simulation: Extracting Signal from Noise",
    template="plotly_white",
    hovermode='x'
)

fig.update_xaxes(title_text="True Temperature (X)", row=1, col=1)
fig.update_yaxes(title_text="Observed Power (Y)", row=1, col=1)
fig.update_xaxes(title_text="Observed Power (Y)", row=1, col=2)
fig.update_yaxes(title_text="Residual Error (Z)", row=1, col=2)

fig.show()
```

The simulation results provide a powerful bridge between abstract probability theory and practical industrial diagnostics. Here is a detailed breakdown of what these results physically and mathematically imply:

1. The Shrinkage Principle (The "Conservative" Estimate)
In the left plot, we compare the **Naive Estimator** ($X=Y$) with the **Optimal Estimator** ($\mathbb{E}[X|Y] = 0.2Y$).
* **The Problem:** Because the noise (Torque $Q$, $\sigma^2=4$) is four times stronger than the signal (Temperature $X$, $\sigma^2=1$), the raw sensor reading $Y$ is mostly "polluted."
* **The Result:** The optimal estimator "shrinks" the measurement by 80%. If the power meter spikes to $Y=5$, the math tells us that, statistically, only $1$ unit of that spike is likely due to temperature, while $4$ units are likely just material hardness.
* **Physical Meaning:** $\mathbb{E}[X|Y]$ prevents **False Alarms**. A naive system would trigger an emergency cooling shut-off every time the tool hits a hard knot in the wood; the "shrunk" estimator recognizes that $Q$ is the more likely culprit for large spikes.

2. The Orthogonality Principle (Efficiency)
The right plot shows the **Residual Error ($Z$)** against the **Observation ($Y$)**.
* **The Observation:** The cloud of points is perfectly horizontal (zero correlation).
* **The Meaning:** This proves that our estimate $0.2Y$ has "sucked out" every possible bit of information that $Y$ had to offer.
* **The Proof:** If the plot showed a diagonal trend, it would imply that some part of the error could still be predicted by $Y$. Since the correlation is zero, the error is purely "white noise" relative to our sensor. We have achieved the **Minimum Mean Square Error (MMSE)**.

3. Information Geometry & Sigma-Algebras
The contour plot visualizes why $X$ is not $\sigma(Y)$-measurable:
* Look at the vertical spread for any single $Y$ value. Even if you fix $Y=0$, $X$ still has a spread (a conditional variance of $0.8$).
* Because the sensor $Y$ maps a 2D reality $(X, Q)$ down to a 1D line, information is lost. The "Best Estimate" is the vertical center-point of that spread, but the "thickness" of the cloud represents the **irreducible uncertainty** ($h(X|Y)$) that the sensor cannot resolve.

4. Variance Reduction and Mutual Information
* **Initial State:** Before measuring, our uncertainty (variance) about the temperature was **1.0**.
* **After Measurement:** Our uncertainty dropped to **0.8**.
* **Information Gain:** We gained **$0.1116$ nats** ($I(X;Y)$).
* **The "Why":** This reduction ($20\%$) is the physical value of the sensor. If we wanted a higher $I(X;Y)$, we wouldn't need a "better" version of $Y$; we would need a **different** sensor (like an acoustic sensor) that is less correlated with $Q$.
